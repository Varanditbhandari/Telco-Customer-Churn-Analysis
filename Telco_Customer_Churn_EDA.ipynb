# Exploratory Data Analysis (EDA) Findings

### 1. Dataset Overview
The dataset contains **7,043 rows and 21 columns**, representing telecommunications customers and their service subscriptions, account information, tenure, monthly charges, total charges, and churn status. The main variables used for the analysis include `tenure`, `MonthlyCharges`, `TotalCharges`, `SeniorCitizen`, and `Churn`.

### 2. Data Quality
The initial inspection revealed no missing values in the original dataset and no duplicate records. However, the `TotalCharges` column was stored as an object data type instead of numeric. Upon conversion to numeric, 11 blank entries were identified and converted to missing values (NaN). The column was successfully converted to a numeric data type to facilitate analysis. The remaining columns were retained in their appropriate data types.

### 3. Numerical Summaries
The descriptive statistics revealed the following:

- **Tenure:** The mean tenure was approximately 32.37 months, while the median was 29 months and the mode was 1 month.
- **Monthly Charges:** The average monthly charge was approximately $64.76, with a median of $70.35 and a mode of $20.05.
- **Total Charges:** The mean total charge was approximately $2,283.30, while the median was $1,397.48 and the mode was $20.20.
- **Senior Citizen:** The mean was approximately 0.16, indicating that about 16.21% of customers were senior citizens. The median and mode were both 0, indicating that the majority of customers were not senior citizens.

### 4. Distributions
The histograms revealed different distribution patterns among the numerical variables. Tenure showed a relatively balanced distribution with a slight positive skew. Monthly charges appeared relatively balanced, with a slight negative skew. Total charges exhibited a noticeable positive skew, indicating that a considerable number of customers had relatively low total charges while fewer customers had substantially higher total charges. The boxplots did not reveal any observations beyond the conventional 1.5×IQR outlier boundaries for the numerical variables examined.

### 5. Churn Distribution
The churn analysis revealed that 1,869 customers (26.54%) had churned, while 5,174 customers (73.46%) remained with the company. This indicates that although the majority of customers were retained, approximately one-quarter of the customer base had discontinued their services, representing a significant area for customer retention improvement.

### 6. Group Comparison
The comparison of numerical variables between churned and retained customers revealed notable differences:

- **Tenure:** Customers who remained with the company had an average tenure of approximately 37.57 months, compared with 17.98 months for churned customers. This suggests that customers with shorter tenure were more likely to churn.
- **Monthly Charges:** Churned customers had a higher average monthly charge of approximately $74.44, compared with $61.27 among retained customers. This indicates a possible relationship between higher monthly charges and customer churn.
- **Total Charges:** Retained customers recorded higher average total charges of approximately $2,555.34, compared with $1,531.80 among churned customers. This may be associated with their longer tenure with the company.
- **Senior Citizen:** The proportion of senior citizens among churned customers was approximately 25.47%, compared with 12.87% among retained customers, indicating that senior citizens represented a larger proportion of the churned customer group.

Overall, the analysis suggests that shorter customer tenure, higher monthly charges, and senior citizen status are associated with higher churn levels. These findings highlight the importance of examining customer tenure, pricing, and customer-specific characteristics when developing strategies to reduce churn and improve customer retention.
